```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow
```

```python
arquivo = files.upload()
```

```python
img_bgr = cv2.imread('rx782.jpg')
img_bgr
```

```python
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
```

```python
img_rgb
```

```python
plt.figure(figsize=(5,3)) #cria um quadro 500x300
plt.imshow(img_rgb)
plt.axis('off')
plt.title('Imagem em RGB')
plt.show()
```

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
```

```python
img_hsv
```

```python
fig, ax = plt.subplots(1, 3, figsize=(6,4))

ax[0].imshow(img_rgb)
ax[0].set_title('RGB')
ax[0].axis('off')

ax[1].imshow(img_bgr)
ax[1].set_title('BGR')
ax[1].axis('off')

ax[2].imshow(img_hsv)
ax[2].set_title('HSV')
ax[2].axis('off')

plt.tight_layout()
plt.show()
```

```python

img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
```

```python
img_gray
```

```python
cv2_imshow(img_gray, cmap = 'gray')
```

```python
fig, ax = plt.subplots(2, 2, figsize=(6,4))

ax[0,0].imshow(img_rgb)
ax[0,0].set_title('RGB')
ax[0,0].axis('off')

ax[0,1].imshow(img_bgr)
ax[0,1].set_title('BGR')
ax[0,1].axis('off')

ax[1,0].imshow(img_hsv)
ax[1,0].set_title('HSV')
ax[1,0].axis('off')

ax[1,1].imshow(img_gray, cmap='gray')
ax[1,1].set_title('Grayscale')
ax[1,1].axis('off')

plt.tight_layout()
plt.savefig('Aula_conversao_de_cores.png')
plt.show()
```
